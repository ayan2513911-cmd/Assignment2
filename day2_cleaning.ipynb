```python
import pandas as pd

data = {
    "OrderID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1002, 1008],
    "Date": ["2025-01-05", "2025-01-06", "2025-01-07", "2025-01-08",
             "2025-01-09", "2025-01-10", "2025-01-11", "2025-01-06", "2025-01-12"],
    "Region": ["North", " south ", "EAST", "West", None, "North", "South", " south ", "North"],
    "Category": ["Electronics", "Furniture", " electronics ", "Office", "Furniture",
                 "Office", "Electronics", "Furniture", "Electronics"],
    "Sales": [120, 80, 200, 50, 60, -10, None, 80, 70],
    "Quantity": [2, 1, 2, 5, 2, 3, 1, 1, 2]
}

df = pd.DataFrame(data)
df.to_csv("sales.csv", index=False)

print(df.head())
df.info()
print(df.describe())
print("Shape:", df.shape)
```

```python
print("Missing Sales:", df["Sales"].isnull().sum())
print("Missing Region:", df["Region"].isnull().sum())
print("Duplicates:", df.duplicated().sum())
print("Regions:", df["Region"].unique())
print("Categories:", df["Category"].unique())
print("Negative Sales:\n", df[df["Sales"] < 0])
```

```python

log = []

n = len(df)
df = df.drop_duplicates()
log.append(["Duplicates", "Removed", f"{n-len(df)} row removed"])

df["Region"] = df["Region"].str.strip().str.title()
df["Category"] = df["Category"].str.strip().str.title()
log.append(["Spaces and capitalization", "Cleaned text", "Text standardized"])

df["Region"] = df["Region"].fillna("Unknown")
log.append(["Missing Region", "Filled with Unknown", "Corrected"])

df["Sales"] = pd.to_numeric(df["Sales"], errors="coerce")
df.loc[df["Sales"] < 0, "Sales"] = None
df["Sales"] = df["Sales"].fillna(df["Sales"].median())
log.append(["Missing and negative Sales", "Filled with median", "Corrected"])

print(df)
print("\nCleaning Log:")
print(pd.DataFrame(log, columns=["Issue", "Correction", "Result"]))

print("\nMissing values:\n", df.isnull().sum())
print("Duplicates:", df.duplicated().sum())
print("Final shape:", df.shape)
```

```python

print(df.groupby("Region")["Sales"].sum())
```

```python

print(pd.pivot_table(
    df,
    values="Sales",
    index="Region",
    columns="Category",
    aggfunc="sum",
    fill_value=0
))
```

Data Quality Summary

1. The dataset contained six data-quality issues, including missing values, duplicates, text errors, and negative sales.
2. Duplicate records were removed, and text spaces and capitalization were corrected.
3. Missing regions were replaced with "Unknown".
4. Missing and negative sales values were replaced using the median.
5. The cleaned dataset contains 8 rows and 6 columns with no missing values or duplicates.
